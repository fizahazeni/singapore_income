```python
#Libraries

from pyspark.sql.functions import col
from pyspark.sql.functions import round
```

```python
#Pull in all 4 tables

df_1 = spark.table("endtoend.iras_data.assessable_income_of_individuals")
df_2 = spark.table("endtoend.iras_data.assessable_income_of_individuals_by_tax_group")
df_3 = spark.table("endtoend.iras_data.income_of_individuals_by_income_type")
df_4 = spark.table("endtoend.iras_data.income_of_individuals_by_income_type_and_tax_group")

```

```python
df_1 = (
    df_1
    .withColumn("mean_total_income", round((col("total_income") / col("no_of_indv_assessed")) * 1000, 2))
    .withColumn("mean_donations", round((col("donations") / col("no_of_indv_assessed")) * 1000, 2))
    .withColumn("mean_assessable_income", round((col("assessable_income") / col("no_of_indv_assessed")) * 1000, 2))
)

df_1.display()
```
